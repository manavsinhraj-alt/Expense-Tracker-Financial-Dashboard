```python
# Cell 2: Core Gateway System Logic (With Offline Mock Fallback)
import sqlite3
import time
import numpy as np
import pandas as pd
from sentence_transformers import SentenceTransformer
from litellm import completion

# 1. Database Setup
def init_db():
    conn = sqlite3.connect("colab_gateway.db")
    cursor = conn.cursor()
    cursor.execute('''
        CREATE TABLE IF NOT EXISTS logs (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            timestamp DATETIME DEFAULT CURRENT_TIMESTAMP,
            prompt TEXT,
            model_used TEXT,
            cached BOOLEAN,
            prompt_tokens INTEGER,
            completion_tokens INTEGER,
            cost_usd REAL,
            latency_ms REAL
        )
    ''')
    conn.commit()
    conn.close()

# 2. Semantic Cache Engine
class SemanticCache:
    def __init__(self, threshold=0.85):
        print("⏳ Loading embedding model...")
        self.model = SentenceTransformer('all-MiniLM-L6-v2')
        self.threshold = threshold
        self.cache = []

    def get(self, prompt: str):
        if not self.cache:
            return None

        prompt_embed = self.model.encode(prompt)
        best_score, best_hit = -1.0, None

        for item in self.cache:
            score = np.dot(prompt_embed, item["embedding"]) / (
                np.linalg.norm(prompt_embed) * np.linalg.norm(item["embedding"])
            )
            if score > best_score:
                best_score = score
                best_hit = item

        if best_score >= self.threshold and best_hit:
            return best_hit["response"], best_hit["model"], best_score
        return None

    def set(self, prompt: str, response: str, model: str):
        prompt_embed = self.model.encode(prompt)
        self.cache.append({
            "embedding": prompt_embed,
            "prompt": prompt,
            "response": response,
            "model": model
        })

# 3. Enterprise Gateway Class
class EnterpriseAIGateway:
    def __init__(self, threshold=0.85):
        init_db()
        self.cache = SemanticCache(threshold=threshold)
        self.fallback_chain = ["gpt-4o-mini", "groq/llama-3.3-70b-versatile"]

        self.pricing = {
            "gpt-4o-mini": {"input": 0.00015, "output": 0.00060},
            "groq/llama-3.3-70b-versatile": {"input": 0.00059, "output": 0.00079},
            "mock-llm-engine": {"input": 0.00010, "output": 0.00020}
        }

    def _calc_cost(self, model, p_tokens, c_tokens):
        p = self.pricing.get(model, {"input": 0.0001, "output": 0.0002})
        return round(((p_tokens/1000.0) * p["input"]) + ((c_tokens/1000.0) * p["output"]), 6)

    def _log_request(self, prompt, model, cached, p_tokens, c_tokens, latency):
        cost = 0.0 if cached else self._calc_cost(model, p_tokens, c_tokens)
        conn = sqlite3.connect("colab_gateway.db")
        cursor = conn.cursor()
        cursor.execute('''
            INSERT INTO logs (prompt, model_used, cached, prompt_tokens, completion_tokens, cost_usd, latency_ms)
            VALUES (?, ?, ?, ?, ?, ?, ?)
        ''', (prompt, model, cached, p_tokens, c_tokens, cost, latency))
        conn.commit()
        conn.close()

    def query(self, prompt: str):
        start = time.time()

        # 1. Check Cache First
        cached_res = self.cache.get(prompt)
        if cached_res:
            res_text, orig_model, score = cached_res
            latency = (time.time() - start) * 1000
            self._log_request(prompt, f"cache-hit ({orig_model})", True, 0, 0, latency)
            return {
                "response": res_text,
                "cached": True,
                "similarity": round(float(score), 4),
                "latency_ms": round(latency, 2),
                "cost_usd": 0.0
            }

        # 2. Try Real API Calls
        for model in self.fallback_chain:
            try:
                res = completion(model=model, messages=[{"role": "user", "content": prompt}], timeout=5)
                latency = (time.time() - start) * 1000
                text = res.choices[0].message.content
                p_tok, c_tok = res.usage.prompt_tokens, res.usage.completion_tokens

                self.cache.set(prompt, text, model)
                self._log_request(prompt, model, False, p_tok, c_tok, latency)

                return {
                    "response": text,
                    "cached": False,
                    "model_used": model,
                    "latency_ms": round(latency, 2),
                    "cost_usd": self._calc_cost(model, p_tok, c_tok)
                }
            except Exception:
                continue

        # 3. Fallback: Mock Response if No API Key is Provided
        time.sleep(0.8) # Simulate 800ms network latency
        model = "mock-llm-engine"
        text = "Transformers rely on Self-Attention mechanism, Positional Encoding, and Feed-Forward Neural Networks."
        p_tok, c_tok = 25, 35
        latency = (time.time() - start) * 1000

        self.cache.set(prompt, text, model)
        self._log_request(prompt, model, False, p_tok, c_tok, latency)

        return {
            "response": text,
            "cached": False,
            "model_used": f"{model} (No API Key Provided)",
            "latency_ms": round(latency, 2),
            "cost_usd": self._calc_cost(model, p_tok, c_tok)
        }

print("✅ Enterprise Gateway Loaded Successfully (Offline/Mock Mode Enabled)!")
```

Security & PII Redaction Engine (app/security.py)

```python
import re

class PIIRedactor:
    def __init__(self):
        # Regex patterns for sensitive enterprise data
        self.patterns = {
            "credit_card": r'\b(?:\d[ -]*?){13,16}\b',
            "email": r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b',
            "api_key": r'(?:sk-|gsk_|key-)[a-zA-Z0-9]{20,}',
            "ssn_aadhaar": r'\b\d{4}[ -]?\d{4}[ -]?\d{4}\b'
        }

    def sanitize(self, text: str) -> tuple[str, dict]:
        """
        Scubs sensitive PII data and returns (sanitized_text, redaction_stats)
        """
        sanitized_text = text
        stats = {}

        for pii_type, pattern in self.patterns.items():
            matches = re.findall(pattern, sanitized_text)
            if matches:
                stats[pii_type] = len(matches)
                sanitized_text = re.sub(pattern, f"[REDACTED_{pii_type.upper()}]", sanitized_text)

        return sanitized_text, stats
```

Step 3: Main Production Gateway Server (app/main.py)

```python
from fastapi import FastAPI, Depends, HTTPException
from pydantic import BaseModel, Field
import time

from app.security import PIIRedactor
from app.auth import verify_api_key
# Re-using our core EnterpriseAIGateway logic
from app.core_gateway import EnterpriseAIGateway

app = FastAPI(
    title="Enterprise AI Gateway & Security Middleware",
    version="2.0.0",
    docs_url="/docs"
)

redactor = PIIRedactor()
gateway = EnterpriseAIGateway(threshold=0.85)

class CompletionRequest(BaseModel):
    prompt: str = Field(..., example="My email is john@company.com and card is 4111-2222-3333-4444. Summarize this document.")
    temperature: float = Field(default=0.7, ge=0.0, le=2.0)

@app.post("/v1/chat/completions", tags=["LLM Gateway"])
async def secure_chat_completion(
    request: CompletionRequest,
    auth_user: dict = Depends(verify_api_key)
):
    start_time = time.time()

    # 1. PII Redaction Layer
    clean_prompt, redactions = redactor.sanitize(request.prompt)

    # 2. Query execution via Gateway (Cache -> Fallback APIs)
    gateway_response = gateway.query(clean_prompt)

    # 3. Inject Security Audit Metadata into Response
    gateway_response["security_audit"] = {
        "client": auth_user["client_name"],
        "pii_detected": len(redactions) > 0,
        "redact_details": redactions,
        "original_prompt_length": len(request.prompt),
        "sanitized_prompt_length": len(clean_prompt)
    }

    return gateway_response

@app.get("/health", tags=["System"])
async def health_check():
    return {"status": "healthy", "engine": "Enterprise AI Gateway v2.0"}
```
